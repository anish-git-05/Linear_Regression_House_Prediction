```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/House_Rent_Dataset.csv')
df.head()
```

```python
df.tail()
```

```python
furnishing_map={"Unfurnished":1,"Semi-Furnished":2,"Furnished":3}
df['Furnishing_val']=df["Furnishing Status"].map(furnishing_map)
```

We have created an extra column which quantifies the furnishing status.

```python
df.head()
```

```python
X=df[['BHK','Size','Bathroom','Furnishing_val']].to_numpy()
```

```python
y=df['Rent'].to_numpy()
y
```

```python
col_mean=np.mean(X,axis=0)
col_sd=np.std(X,axis=0)
```

```python
X_scaled=(X-col_mean)/col_sd
```

```python
print(np.mean(X_scaled,axis=0))
print(np.std(X_scaled,axis=0))
```

The mean for all columns is close to 0 and the standard deviation is close to 1.

```python
w=np.zeros(X_scaled.shape[1])
b=0.0
initial_loss=np.mean((X_scaled@w+b-y)**2)
def gradient_descent(X,y,w,b,alpha=0.01,iters=1000):
  n=X.shape[0]
  for i in range(iters):
    y_cap=X@w+b
    error=y_cap-y
    dw=(2/n)*(X.T@error)
    db=(2/n)*(np.sum(error))
    w=w-alpha*dw
    b=b-alpha*db
  return w,b

w_new,b_new=gradient_descent(X_scaled,y,w,b,alpha=0.01,iters=1000)
print(w_new)
print(b_new)
```

```python
final_loss=np.mean((X_scaled@w_new+b_new)**2)
print(np.sqrt(initial_loss))
print(np.sqrt(final_loss))
```

```python
print(X_scaled@w_new+b_new)
```

We have to take into account the city of rent. So we have done the binary encoding of each city present in the data.

```python
df['Is_Kolkata']=(df['City']=='Kolkata').astype(int)
df['Is_Delhi']=(df['City']=='Delhi').astype(int)
df['Is_Mumbai']=(df['City']=='Mumbai').astype(int)
df['Is_Chennai']=(df['City']=='Chennai').astype(int)
df['Is_Bangalore']=(df['City']=='Bangalore').astype(int)
df['Is_Hyderabad']=(df['City']=='Hyderabad').astype(int)
X_num=df[['BHK','Size','Bathroom','Furnishing_val']]
X_city=df[['Is_Kolkata','Is_Delhi','Is_Mumbai','Is_Chennai','Is_Bangalore','Is_Hyderabad']].to_numpy()
```

```python
col_mean=np.mean(X_num,axis=0)
col_sd=np.std(X_num,axis=0)
X_scaled=(X_num-col_mean)/(col_sd)
X_final=np.concatenate((X_scaled,X_city),axis=1)
```

Only scaled the numeical continuous values. Didn't scale the discrete binary indicators.

```python
w=np.zeros(X_final.shape[1])
b=0.0
w_new,b_new=gradient_descent(X_final,y,w,b,alpha=0.001)
new_loss=np.mean((X_final@w_new+b_new-y)**2)
print(np.sqrt(new_loss))
print(np.mean(y))
y_pred=X_final@w_new+b_new
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
lr=LinearRegression()
lr.fit(X_final,y)
y_pred=lr.predict(X_final)
print(np.sqrt(mean_squared_error(y,y_pred)))
print(np.mean(y))
```

We can see there is not much difference between the Root mean square error of gradient descent approach and by using the built in function of sklearn library.
